# Exploración de FIFA Ranking

Carga y revisión inicial del dataset `fifa_ranking.csv`.

```python
from pathlib import Path
import pandas as pd

dataset_candidates = [Path('../fifa_ranking.csv'), Path('fifa_ranking.csv')]
dataset_path = next((path for path in dataset_candidates if path.exists()), None)
if dataset_path is None:
    raise FileNotFoundError('No se encontró fifa_ranking.csv')

df = pd.read_csv(dataset_path)
print(f'Dataset cargado desde: {dataset_path.resolve()}')
print(f'Registros: {df.shape[0]:,} | Columnas: {df.shape[1]}')
```

```python
df.info()
```

```python
pd.DataFrame({
    'tipo': df.dtypes.astype(str),
    'valores_nulos': df.isna().sum(),
    'valores_unicos': df.nunique(),
})
```

```python
df.describe(include='all').T
```

```python
df['rank_date'] = pd.to_datetime(df['rank_date'])
print('Periodo:', df['rank_date'].min().date(), 'a', df['rank_date'].max().date())
df.sort_values(['rank_date', 'rank']).head(10)
```
